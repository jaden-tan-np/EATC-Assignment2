# FraudWatch

## Import packages

```python
import numpy as np
import joblib
import pandas as pd
import streamlit as st
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
)
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.svm import SVC
from xgboost import XGBClassifier
```

## Load and combine datasets

```python
# Filepaths
train_path = "fraudTrain.csv"
test_path = "fraudTest.csv"

# Load the files
df_train = pd.read_csv(train_path, index_col=0)
df_test = pd.read_csv(test_path, index_col=0)

# Combine into single dataframe
df = pd.concat([df_train, df_test], axis=0, ignore_index=True)

# The full dataset contains over 1.8 million records, which would slow down training and
# testing greatly to process in full. For the sake of our assignment, we will be only
# be taking the first 150,000 rows of the test dataset.

# Take a sample of 150,000 rows
df = df.sample(20000)
df.head()
```

## Data Cleaning


```python
df["merchant"] = df["merchant"].apply(lambda x: str(x).replace("fraud_", ""))
```

```python
# Drop unused text/unique string identifiers to prevent data leakage
drop_cols = ["trans_num", "first", "last", "street", "city", "state", "zip", "dob", "cc_num"]
df_clean = df.drop(columns=[col for col in drop_cols if col in df.columns])

# Convert timestamp to temporal features
if "trans_date_trans_time" in df_clean.columns:
    df_clean["trans_date_trans_time"] = pd.to_datetime(
        df_clean["trans_date_trans_time"]
    )
    df_clean["hour"] = df_clean["trans_date_trans_time"].dt.hour
    df_clean["day_of_week"] = df_clean["trans_date_trans_time"].dt.dayofweek
    df_clean.drop(columns=["trans_date_trans_time"], inplace=True)
```

```python
# Encode categorical variables like job or gender
label_encoders = {}
categorical_cols = ["merchant", "category", "gender", "job"]

for col in categorical_cols:
    if col in df_clean.columns:
        le = LabelEncoder()

        # Get values as strings
        values = df_clean[col].astype(str).values

        # Add "Unknown" as an extra class
        values_with_unknown = np.append(values, "Unknown")

        # Fit the encoder
        le.fit(values_with_unknown)

        # Transform the original data
        df_clean[col] = le.transform(values)

        # Save the encoder
        label_encoders[col] = le

df.head()
```

```python
col = "job"
le = label_encoders[col]

# Convert to string
df_clean[col] = df_clean[col].astype(str)

# Replace unseen values with "Unknown"
df_clean[col] = df_clean[col].apply(
    lambda x: x if x in le.classes_ else "Unknown"
)

# Transform
df_clean[col] = le.transform(df_clean[col])

print(label_encoders.keys())
```

```python
# Define Features (X) and Target (y)
X = df_clean.drop(columns=["is_fraud"])
y = df_clean["is_fraud"]
```

## Test/Train Split


```python
# Stratification ensures identical fraud ratios in train and test splits
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y)
```

```python
# Feature Scaling for SVC distance calculations
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Calculate ratio for XGBoost class weighting
ratio = (len(y_train) - sum(y_train)) / sum(y_train)
```

## Model Training

```python
# SVC requires scaled features and stratified sub-sampling if dataset is massive
# CalibratedClassifierCV enables probability outputs for ROC-AUC scoring & Streamlit confidence meters
print("--- Training Model 1: Support Vector Classifier (SVC) ---")
svc_model = CalibratedClassifierCV(
    estimator=SVC(
        kernel="rbf",
        C=1.0,
        class_weight="balanced",
    ),
    ensemble=False,
)

# For standard execution, fit on scaled training set:
svc_model.fit(X_train_scaled, y_train)
```

## Train XGBoost separately

```python
print("--- Training Model 2: XGBoost Classifier ---")
xgb_model = XGBClassifier(
    n_estimators=100,
    scale_pos_weight=ratio,
    learning_rate=0.1,
    max_depth=6,
    n_jobs=-1,
)
xgb_model.fit(X_train_scaled, y_train)
```

## Evaluate Models

```python
# Function to evaluate each model
def evaluate_model(model, name, X_test_scaled, y_test):
    y_pred = model.predict(X_test_scaled)
    y_proba = model.predict_proba(X_test_scaled)[:, 1]

    metrics = {
        "Model": name,
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_proba),
    }

    print(f"---------- {name} Evaluation Report ----------")
    print(classification_report(y_test, y_pred, digits=4))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    return metrics
```

```python
svc_results = evaluate_model(
    svc_model, "Support Vector Classifier (SVC)", X_test_scaled, y_test
)
xgb_results = evaluate_model(xgb_model, "XGBoost", X_test_scaled, y_test)

# Display side-by-side comparison
comparison_df = pd.DataFrame([svc_results, xgb_results])
print("---------- MODEL COMPARISON SUMMARY ----------")
print(comparison_df.to_string(index=False))
```

```python
xgb_results = evaluate_model(xgb_model, "XGBoost", X_test_scaled, y_test)

# Display side-by-side comparison
comparison_df = pd.DataFrame([xgb_results])
print("---------- MODEL COMPARISON SUMMARY ----------")
print(comparison_df.to_string(index=False))
```

```python
# Save artifacts for streamlit app
print("\nSaving trained models and preprocessing artifacts...")
joblib.dump(svc_model, "svc_model.pkl")
joblib.dump(scaler, "scaler.pkl")

# Save the trained XGBoost model
joblib.dump(xgb_model, "xgb_fraud_model.pkl")

# Save label encoders
joblib.dump(label_encoders, "label_encoders.pkl")

# Save exact feature names and column order used during training
joblib.dump(X_train.columns.tolist(), "model_features.pkl")


joblib.dump(X.columns.tolist(), "feature_names.pkl")
```

# Optimized

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.metrics import classification_report, confusion_matrix
import category_encoders as ce
```

```python
# Filepaths
train_path = "fraudTrain.csv"
test_path = "fraudTest.csv"

# Load the files
df_train = pd.read_csv(train_path, index_col=0)
df_test = pd.read_csv(test_path, index_col=0)
df = pd.concat([df_train, df_test], axis=0, ignore_index=True)

# Combine into single dataframe
df = pd.concat([df_train, df_test], axis=0, ignore_index=True)

# Separate the dataset by class
fraud_cases = df[df['is_fraud'] == 1]
legit_cases = df[df['is_fraud'] == 0]

# Keep ALL fraud cases, and sample legitimate cases to reach your desired dataset size
# (Assuming fraud_cases is around 10,000 rows, we sample 90,000 legit cases)
legit_sample = legit_cases.sample(n=90000)

# Combine and shuffle the new balanced-but-smaller dataset
df = pd.concat([fraud_cases, legit_sample]).sample(frac=1, random_state=42).reset_index(drop=True)
```

## Feature Engineering

```python
# Convert date columns to datetime objects
df['trans_date'] = pd.to_datetime(df['trans_date_trans_time'])
df['dob'] = pd.to_datetime(df['dob'])

# A. Calculate Customer Age at the time of transaction
df['customer_age'] = (df['trans_date'] - df['dob']).dt.days // 365

# B. Calculate Haversine Distance between customer and merchant
def haversine_array(lat1, lon1, lat2, lon2):
    R = 6371.0 # Earth radius in kilometers
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2.0)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

df['distance_to_merchant'] = haversine_array(
    df['lat'], df['long'], 
    df['merch_lat'], df['merch_long']
)

# C. Calculate Transaction Velocity (Transactions in the last 24 hours per credit card)
df = df.sort_values(by=['cc_num', 'trans_date'])
df.set_index('trans_date', inplace=True)
df['trans_count_24h'] = df.groupby('cc_num')['amt'].transform(lambda x: x.rolling('24H').count())
df.reset_index(inplace=True)

# Drop the columns that have now been engineered or are prone to leakage
cols_to_drop = [
    'trans_date_trans_time', 'trans_date', 'dob', 'cc_num', 
    'lat', 'long', 'merch_lat', 'merch_long', 'first', 'last', 'street', 'city', 'state', 'zip', 'trans_num'
]
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])
```

## Train-Test Split


```python
# Separate features (X) and target (y)
X = df.drop(columns=['is_fraud'])
y = df['is_fraud']

# Implement an 80/20 train-test split for model validation
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y  
)
```

## Categorical Encoding

```python
# A. One-Hot Encoding for low-cardinality features
low_cardinality_cols = ['gender', 'category']
X_train = pd.get_dummies(X_train, columns=low_cardinality_cols, drop_first=True)
X_test = pd.get_dummies(X_test, columns=low_cardinality_cols, drop_first=True)

# Ensure both train and test have the same dummy columns after encoding
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)

# B. Target Encoding for high-cardinality features
high_cardinality_cols = ['merchant', 'job']
target_encoder = ce.TargetEncoder(cols=high_cardinality_cols, smoothing=10)

# Fit on training data ONLY to prevent data leakage, then transform both
X_train = target_encoder.fit_transform(X_train, y_train)
X_test = target_encoder.transform(X_test)
```

## Model Setup & Hyperparemeter tuning

```python
# Calculate scale_pos_weight to handle the severe class imbalance
fraud_ratio = y_train.value_counts()[0] / y_train.value_counts()[1]

xgb_model = xgb.XGBClassifier(
    objective='binary:logistic',
    scale_pos_weight=fraud_ratio,
    random_state=42,
    eval_metric='auc'
)

# Define the parameter grid for tuning
param_grid = {
    'max_depth': [4, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'n_estimators': [100, 200, 300],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform Randomized Search
random_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_grid,
    n_iter=10,        
    scoring='f1',     
    cv=3,             
    verbose=1,
    random_state=42,
    n_jobs=-1         
)

print("Starting Hyperparameter Tuning...")
random_search.fit(X_train, y_train)
```

## Model Eval

```python
print("Best Parameters Found:")
print(random_search.best_params_)

# Evaluate the best model on the unseen test set
best_model = random_search.best_estimator_
y_pred = best_model.predict(X_test)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
